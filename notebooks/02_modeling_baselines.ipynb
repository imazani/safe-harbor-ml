```python
from pathlib import Path
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score, roc_auc_score, precision_recall_curve
from sklearn.ensemble import RandomForestClassifier


PROJECT_ROOT = Path("..").resolve()
DATA_PROCESSED = PROJECT_ROOT / "data" / "processed"

df = pd.read_csv(DATA_PROCESSED / "features.csv")
df.shape, df["label"].value_counts()
```

### Define feature matrix x and target y

```python
y = df["label"].astype(int)

feature_cols = [c for c in df.columns if c.endswith("__mean") or c.endswith("__max") or c.endswith("__coverage")]
X = df[feature_cols].copy()

meta_cols = ["Chromosome", "Start", "End", "locus", "tier"]
meta = df[[c for c in meta_cols if c in df.columns]].copy()

X.shape, len(feature_cols), feature_cols
```

### Regularized logostic regression

```python
logreg = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="constant", fill_value=0.0)),  # keep coverage signal; treat missing as 0
        ("scaler", StandardScaler()),
        ("clf", LogisticRegression(
            penalty="l2",
            C=1.0,
            class_weight="balanced",
            solver="liblinear",
            max_iter=500,
            random_state=42
        ))
    ]
)

logreg
```

### Leave-One-Positive-Out evaluation

```python
pos_idx = np.where(y.values == 1)[0]
neg_idx = np.where(y.values == 0)[0]

rng = np.random.default_rng(42)

def precision_at_k(y_true, y_score, k):
    order = np.argsort(-y_score)
    topk = y_true[order][:k]
    return topk.mean() if k > 0 else np.nan

results = []

for fold, held_pos in enumerate(pos_idx, start=1):
    test_pos = np.array([held_pos])
    
    # Choose a manageable number of negatives for test to make PR curves meaningful
    test_neg = rng.choice(neg_idx, size=300, replace=False)
    test_idx = np.concatenate([test_pos, test_neg])
    
    # Train uses the other positives + a larger negative sample
    train_pos = np.array([i for i in pos_idx if i != held_pos])
    remaining_negs = np.setdiff1d(neg_idx, test_neg, assume_unique=False)
    train_neg = rng.choice(remaining_negs, size=600, replace=False)
    train_idx = np.concatenate([train_pos, train_neg])
    
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_test, y_test = X.iloc[test_idx], y.iloc[test_idx]
    
    logreg.fit(X_train, y_train)
    y_score = logreg.predict_proba(X_test)[:, 1]
    
    ap = average_precision_score(y_test, y_score)
    roc = roc_auc_score(y_test, y_score)
    
    p_at_10 = precision_at_k(y_test.values, y_score, 10)
    p_at_25 = precision_at_k(y_test.values, y_score, 25)
    p_at_50 = precision_at_k(y_test.values, y_score, 50)
    
    results.append({
        "fold": fold,
        "held_out_positive_index": int(held_pos),
        "AP(PR-AUC)": ap,
        "ROC-AUC": roc,
        "Precision@10": p_at_10,
        "Precision@25": p_at_25,
        "Precision@50": p_at_50,
    })

results_df = pd.DataFrame(results)
results_df
```

Interpretation notes:

PR-AUC is the main metric here

Precision@K tells you: “If I look at the top K predicted windows, how many are actually positive?”

With only 1 positive in each test fold, Precision@K will be either 0 or 1/K — that’s expected. The trend across folds matters more than any single value.

Inspect which features drive predictions using coefficients

```python
# Pull trained coefficients from the last fold fit
clf = logreg.named_steps["clf"]
scaler = logreg.named_steps["scaler"]
imputer = logreg.named_steps["imputer"]

coef = clf.coef_.ravel()
coef_df = pd.DataFrame({"feature": feature_cols, "coef": coef}).sort_values("coef", ascending=False)

coef_df.head(15), coef_df.tail(15)

top_pos = coef_df.head(10)
top_neg = coef_df.tail(10)

pd.concat([top_pos.assign(direction="pushes positive"),
           top_neg.assign(direction="pushes negative")], ignore_index=True)
```

```python
### Simple non-linear model (random-forest) for comparison
```

```python
rf = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="constant", fill_value=0.0)),
        ("clf", RandomForestClassifier(
            n_estimators=500,
            random_state=42,
            class_weight="balanced_subsample",
            min_samples_leaf=5
        ))
    ]
)

rf_results = []

for fold, held_pos in enumerate(pos_idx, start=1):
    test_pos = np.array([held_pos])
    test_neg = rng.choice(neg_idx, size=300, replace=False)
    test_idx = np.concatenate([test_pos, test_neg])

    train_pos = np.array([i for i in pos_idx if i != held_pos])
    remaining_negs = np.setdiff1d(neg_idx, test_neg, assume_unique=False)
    train_neg = rng.choice(remaining_negs, size=600, replace=False)
    train_idx = np.concatenate([train_pos, train_neg])

    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_test, y_test = X.iloc[test_idx], y.iloc[test_idx]

    rf.fit(X_train, y_train)
    y_score = rf.predict_proba(X_test)[:, 1]

    rf_results.append({
        "fold": fold,
        "AP(PR-AUC)": average_precision_score(y_test, y_score),
        "ROC-AUC": roc_auc_score(y_test, y_score),
        "Precision@10": precision_at_k(y_test.values, y_score, 10),
        "Precision@25": precision_at_k(y_test.values, y_score, 25),
        "Precision@50": precision_at_k(y_test.values, y_score, 50),
    })

pd.DataFrame(rf_results)
```
